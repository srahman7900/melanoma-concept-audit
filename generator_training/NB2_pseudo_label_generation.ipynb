```python
# ============================================================
# NOTEBOOK 2 — Pseudo-Label Generation for ISIC 2020 via CBM-A
# Single-cell version. Simple local Kaggle save (no Drive).
# Platform: Kaggle T4×2 GPU
#
# ⚠ BEFORE RUNNING — VERIFY:
#   1. ISIC_IMG_DIR — confirmed from earlier project work
#   2. ISIC_GT_CSV — ⚠ BEST GUESS, not confirmed. If wrong, the code
#      lists every CSV found under the dataset root so you can fix
#      the path in one round-trip instead of guessing blindly.
#   3. CBM_A_CHECKPOINT — path to best_cbm_model.pth from NB1
#      (upload it as a new Kaggle dataset first, or use /kaggle/working/
#      if running in the same session as NB1)
#
# WHAT THIS PRODUCES:
#   isic2020_pseudo_labels_v3.csv — image_name, target (real ISIC label),
#   concept_<name> ×7 (CBM-A predicted probabilities)
#   This is the PSEUDO_CSV that NB3/NB3b/NB4 will load.
#
# RESUMABLE: if interrupted, rerun — already-processed images are
# detected from the partial output CSV and skipped.
# ============================================================

# ── INSTALL ────────────────────────────────────────────────────
import subprocess, sys
for pkg in ["timm==1.0.15"]:
    subprocess.run([sys.executable, '-m', 'pip', 'install', pkg, '-q'], check=False)

import os, time, glob, warnings
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torchvision.models as tv_models
from torchvision import transforms
from PIL import Image
from torch.utils.data import Dataset, DataLoader
import matplotlib.pyplot as plt
from tqdm import tqdm
warnings.filterwarnings('ignore')

device = torch.device('cuda')
torch.backends.cudnn.benchmark = True
print(f"GPU: {torch.cuda.get_device_name(0)} | Count: {torch.cuda.device_count()}")


# ============================================================
# CONFIG — verify these paths
# ============================================================

ISIC_IMG_DIR = "/kaggle/input/datasets/sumaiyabinteshahid/isic-challenge-dataset-2020/ISIC_2020_Dataset/train"

# ⚠ BEST GUESS — standard ISIC 2020 challenge filename. If this fails,
# the code below lists every CSV under the dataset root automatically.
ISIC_GT_CSV = "/kaggle/input/datasets/sumaiyabinteshahid/isic-challenge-dataset-2020/ISIC_2020_Dataset/ISIC_2020_Train_Metadata.csv"

# ⚠ Path to best_cbm_model.pth from NB1 — update after uploading as a dataset
CBM_A_CHECKPOINT = "/kaggle/input/datasets/shohanurrahman7/best-cbm-model-128/best_cbm_model.pth"

OUTPUT_CSV  = 'isic2020_pseudo_labels_v3.csv'
IMAGE_SIZE  = 224          # MUST match CBM-A's training resolution (NB1)
BATCH_SIZE  = 64           # inference-only, no backward pass — larger batch OK
NUM_WORKERS = 4
SAVE_EVERY_N_BATCHES = 50  # incremental write frequency

CONCEPT_NAMES = ['pigment_network', 'blue_whitish_veil', 'vascular_structures',
                  'pigmentation', 'streaks', 'dots_and_globules', 'regression_structures']
NUM_CONCEPTS  = len(CONCEPT_NAMES)


# ============================================================
# PATH VERIFICATION — fail loud with actionable info, not silently
# ============================================================

if not os.path.exists(ISIC_IMG_DIR):
    raise FileNotFoundError(f"⚠ ISIC image directory not found: {ISIC_IMG_DIR}")

n_images_in_dir = len([f for f in os.listdir(ISIC_IMG_DIR) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
print(f"✓ ISIC image dir found: {n_images_in_dir} image files present")

if not os.path.exists(ISIC_GT_CSV):
    print(f"\n⚠ ISIC_GT_CSV not found at guessed path: {ISIC_GT_CSV}")
    print(f"⚠ Searching for CSV files under the dataset root to help locate the correct one...")
    dataset_root = "/kaggle/input/datasets/sumaiyabinteshahid/isic-challenge-dataset-2020"
    found_csvs = glob.glob(os.path.join(dataset_root, '**', '*.csv'), recursive=True)
    if found_csvs:
        print(f"  Found {len(found_csvs)} CSV file(s):")
        for f in found_csvs:
            print(f"    {f}")
        print(f"\n  Update ISIC_GT_CSV above to the correct path from this list, then rerun.")
    else:
        print(f"  No CSV files found under {dataset_root} — check the dataset is attached correctly.")
    raise FileNotFoundError(
        f"⚠ Stopping here — fix ISIC_GT_CSV using the list above before proceeding."
    )

gt_df = pd.read_csv(ISIC_GT_CSV)
print(f"\n✓ ISIC ground truth CSV loaded: {len(gt_df)} rows. Columns: {list(gt_df.columns)}")

if 'image_name' not in gt_df.columns or 'target' not in gt_df.columns:
    raise KeyError(
        f"⚠ Expected columns 'image_name' and 'target' not both found in {list(gt_df.columns)}.\n"
        f"Update the column references below to match your actual CSV."
    )

print(f"  Target distribution: {gt_df['target'].value_counts().to_dict()} (1=malignant/melanoma, 0=benign)")

if not os.path.exists(CBM_A_CHECKPOINT):
    raise FileNotFoundError(
        f"⚠ CBM-A checkpoint not found at {CBM_A_CHECKPOINT}\n"
        f"Upload best_cbm_model.pth from NB1 as a new Kaggle dataset and update this path."
    )


# ============================================================
# MODEL — CBM-A (must exactly match NB1's architecture, loaded strict=True
# as a built-in consistency check between NB1 and NB2)
# ============================================================

class CBM_A(nn.Module):
    def __init__(self, num_concepts=7):
        super().__init__()
        bb = tv_models.efficientnet_v2_s(weights=None)   # weights loaded from checkpoint below
        in_f = bb.classifier[1].in_features
        bb.classifier = nn.Identity()
        self.backbone = bb
        self.concept_feature = nn.Sequential(
            nn.Linear(in_f, 512), nn.LayerNorm(512), nn.GELU(), nn.Dropout(0.3)
        )
        self.concept_heads = nn.ModuleList([
            nn.Sequential(nn.Linear(512, 128), nn.GELU(), nn.Dropout(0.2), nn.Linear(128, 1))
            for _ in range(num_concepts)
        ])
        self.diagnosis_head = nn.Sequential(
            nn.Linear(num_concepts, 64), nn.GELU(), nn.Dropout(0.3), nn.Linear(64, 2)
        )

    def forward(self, x):
        f = self.backbone(x)
        if f.dim() > 2:
            f = f.mean([-2, -1])
        cf = self.concept_feature(f)
        cl = torch.cat([h(cf) for h in self.concept_heads], dim=1)
        cp = torch.sigmoid(cl)
        diag_logits = self.diagnosis_head(cp)
        return cl, cp, diag_logits


model = CBM_A(num_concepts=NUM_CONCEPTS)
ck = torch.load(CBM_A_CHECKPOINT, map_location='cpu', weights_only=False)
model.load_state_dict(ck['model_state_dict'], strict=True)   # strict=True: architecture mismatch fails loudly
model = model.to(device).eval()
print(f"\n✓ CBM-A loaded from NB1 checkpoint (epoch {ck.get('epoch', '?')})")

if torch.cuda.device_count() > 1:
    print(f"Using {torch.cuda.device_count()} GPUs via DataParallel")
    model = nn.DataParallel(model)


# ============================================================
# RESOLVE IMAGE PATHS
# ============================================================

def resolve_path(name):
    name = str(name)
    for ext in ['', '.jpg', '.jpeg', '.png']:
        p = os.path.join(ISIC_IMG_DIR, name + ext)
        if os.path.exists(p):
            return p
    return None

gt_df['image_path'] = gt_df['image_name'].apply(resolve_path)
found = gt_df['image_path'].notna().sum()
print(f"\nImage resolution: {found}/{len(gt_df)} found at {ISIC_IMG_DIR}")
if found == 0:
    raise FileNotFoundError(f"⚠ No images found. Check ISIC_IMG_DIR and image_name format.")
gt_df = gt_df[gt_df['image_path'].notna()].reset_index(drop=True)


# ============================================================
# RESUMABILITY — skip already-processed images if rerunning
# ============================================================

already_done = set()
if os.path.exists(OUTPUT_CSV):
    prev_df = pd.read_csv(OUTPUT_CSV)
    already_done = set(prev_df['image_name'])
    print(f"\n✓ Found existing partial output: {len(already_done)} images already processed — will skip these.")
else:
    prev_df = None
    print(f"\nFresh start — no existing output found.")

remaining_df = gt_df[~gt_df['image_name'].isin(already_done)].reset_index(drop=True)
print(f"Remaining to process: {len(remaining_df)}/{len(gt_df)}")

if len(remaining_df) == 0:
    print("\n✓ All images already processed. Skipping inference, jumping to summary.")


# ============================================================
# DATASET / DATALOADER (deterministic — no augmentation, this is inference)
# ============================================================

class InferenceDataset(Dataset):
    def __init__(self, df, image_size):
        self.df = df.reset_index(drop=True)
        self.tfm = transforms.Compose([
            transforms.Resize((image_size, image_size)),
            transforms.ToTensor(),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
        ])

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        try:
            img = self.tfm(Image.open(row['image_path']).convert('RGB'))
        except Exception:
            img = torch.zeros(3, 224, 224)
        return img, row['image_name'], int(row['target'])


if len(remaining_df) > 0:
    infer_ds = InferenceDataset(remaining_df, IMAGE_SIZE)
    infer_loader = DataLoader(infer_ds, batch_size=BATCH_SIZE, shuffle=False,
                              num_workers=NUM_WORKERS, pin_memory=True)

    t0 = time.time()
    _ = next(iter(infer_loader))
    batch_time = time.time() - t0
    n_batches = len(infer_loader)
    est_min = batch_time * n_batches / 60
    print(f"~{batch_time:.2f}s/batch → {n_batches} batches ≈ {est_min:.1f} min total")


    # ============================================================
    # INFERENCE LOOP — incremental save every SAVE_EVERY_N_BATCHES
    # ============================================================

    print(f"\n{'='*60}")
    print(f"RUNNING INFERENCE — {len(remaining_df)} images")
    print(f"{'='*60}\n")

    rows_buffer = []

    def flush_buffer(rows_buffer, output_csv):
        if not rows_buffer:
            return
        new_df = pd.DataFrame(rows_buffer)
        if os.path.exists(output_csv):
            new_df.to_csv(output_csv, mode='a', header=False, index=False)
        else:
            new_df.to_csv(output_csv, mode='w', header=True, index=False)
        rows_buffer.clear()

    t_start = time.time()
    with torch.no_grad():
        for batch_idx, (imgs, names, targets) in enumerate(tqdm(infer_loader, desc='inference')):
            imgs = imgs.to(device, non_blocking=True)
            _, cp, _ = model(imgs)
            cp = cp.cpu().numpy()

            for i in range(len(names)):
                row = {'image_name': names[i], 'target': int(targets[i])}
                for ci, cname in enumerate(CONCEPT_NAMES):
                    row[f'concept_{cname}'] = float(cp[i, ci])
                rows_buffer.append(row)

            if (batch_idx + 1) % SAVE_EVERY_N_BATCHES == 0:
                flush_buffer(rows_buffer, OUTPUT_CSV)
                elapsed = time.time() - t_start
                done = (batch_idx + 1) * BATCH_SIZE
                eta_min = (elapsed / (batch_idx + 1)) * (n_batches - batch_idx - 1) / 60
                print(f"  [checkpoint] {min(done, len(remaining_df))}/{len(remaining_df)} processed, "
                      f"saved to {OUTPUT_CSV} | ETA={eta_min:.1f}min")

    flush_buffer(rows_buffer, OUTPUT_CSV)
    print(f"\n✓ Inference complete. Total time: {(time.time()-t_start)/60:.1f} min")
else:
    print("(Skipped — already complete)")


# ============================================================
# FINAL VERIFICATION + SUMMARY
# ============================================================

final_df = pd.read_csv(OUTPUT_CSV)
final_df = final_df.drop_duplicates(subset='image_name', keep='last').reset_index(drop=True)
final_df.to_csv(OUTPUT_CSV, index=False)   # rewrite deduplicated

print(f"\n{'='*60}")
print(f"FINAL PSEUDO-LABEL SUMMARY")
print(f"{'='*60}")
print(f"Total images processed: {len(final_df)}")
print(f"Target distribution: {final_df['target'].value_counts().to_dict()}")

concept_cols = [f'concept_{c}' for c in CONCEPT_NAMES]
mel_df = final_df[final_df['target'] == 1]
ben_df = final_df[final_df['target'] == 0]

print(f"\nPer-concept mean (melanoma n={len(mel_df)} vs benign n={len(ben_df)}):")
print(f"  {'Concept':<24} {'Melanoma':>10} {'Benign':>10} {'Diff':>8}")
print(f"  {'-'*54}")
empirical_mel_vec, empirical_ben_vec = [], []
for c in CONCEPT_NAMES:
    col = f'concept_{c}'
    m_mean = mel_df[col].mean()
    b_mean = ben_df[col].mean()
    empirical_mel_vec.append(m_mean)
    empirical_ben_vec.append(b_mean)
    print(f"  {c:<24} {m_mean:>10.4f} {b_mean:>10.4f} {m_mean-b_mean:>8.4f}")

print(f"\n⚠ EMPIRICAL conditioning vectors derived from this corpus "
      f"(vs hand-set v1/v2 vectors):")
print(f"  empirical_MEL_VEC = {[round(v,3) for v in empirical_mel_vec]}")
print(f"  empirical_BEN_VEC = {[round(v,3) for v in empirical_ben_vec]}")
print(f"\n  Old hand-set MEL_VEC = [0.85, 0.45, 0.20, 0.75, 0.65, 0.82, 0.18]")
print(f"  Old hand-set BEN_VEC = [0.25, 0.02, 0.08, 0.28, 0.12, 0.35, 0.02]")

old_mel = np.array([0.85, 0.45, 0.20, 0.75, 0.65, 0.82, 0.18])
old_ben = np.array([0.25, 0.02, 0.08, 0.28, 0.12, 0.35, 0.02])
emp_mel = np.array(empirical_mel_vec)
emp_ben = np.array(empirical_ben_vec)
cos_mel = np.dot(old_mel, emp_mel) / (np.linalg.norm(old_mel) * np.linalg.norm(emp_mel) + 1e-8)
cos_ben = np.dot(old_ben, emp_ben) / (np.linalg.norm(old_ben) * np.linalg.norm(emp_ben) + 1e-8)
print(f"\n  Cosine similarity (old hand-set vs new empirical): "
      f"mel={cos_mel:.4f} | ben={cos_ben:.4f}")
print(f"  → Recommendation: use the EMPIRICAL vectors in NB3 — they reflect")
print(f"    this specific CBM-A's actual learned concept space, and are more")
print(f"    defensible in the paper than hand-picked values.")


# ============================================================
# FIGURE — concept distributions by class (PNG + PDF)
# ============================================================

def save_fig_both(fig, name, dpi=300):
    fig.savefig(f'{name}.png', dpi=dpi, bbox_inches='tight', facecolor='white')
    fig.savefig(f'{name}.pdf', bbox_inches='tight', facecolor='white')
    print(f"  Saved: {name}.png + {name}.pdf")


fig, axes = plt.subplots(2, 4, figsize=(18, 8))
axes = axes.flatten()
for i, c in enumerate(CONCEPT_NAMES):
    col = f'concept_{c}'
    axes[i].hist(ben_df[col], bins=30, alpha=0.6, label='Benign', color='#2980b9', density=True)
    axes[i].hist(mel_df[col], bins=30, alpha=0.6, label='Melanoma', color='#c0392b', density=True)
    axes[i].set_title(c, fontsize=10)
    axes[i].legend(fontsize=8)
    axes[i].grid(alpha=0.3)
axes[7].axis('off')   # 8th panel unused (7 concepts)

plt.suptitle('CBM-A Pseudo-Label Distributions by True Diagnosis (ISIC 2020)', fontweight='bold')
plt.tight_layout()
save_fig_both(fig, 'nb2_pseudo_label_distributions')
plt.show()

print(f"\n{'='*60}")
print(f"Saved: {OUTPUT_CSV}")
print(f"This is the PSEUDO_CSV that NB3/NB3b/NB4 will load.")
print(f"{'='*60}")
```

```python
# ============================================================
# NB2b — VISUAL SANITY CHECK: Real images at concept-score extremes
# No GPU needed — pure CPU, image loading + matplotlib only.
# Purpose: visually verify whether CBM-A's blue_whitish_veil collapse
# reflects something real in the images, or looks like a model failure,
# BEFORE committing the 128×128 budget to NB3.
# ============================================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ⚠ Update if running in a fresh session (not the same one that ran NB2)
PSEUDO_CSV   = "/kaggle/input/datasets/shohanurrahman7/isic2020-pseudo-labels-v3/isic2020_pseudo_labels_v3.csv"   # or wherever you uploaded it
ISIC_IMG_DIR = "/kaggle/input/datasets/sumaiyabinteshahid/isic-challenge-dataset-2020/ISIC_2020_Dataset/train"

df = pd.read_csv(PSEUDO_CSV)
print(f"Loaded {len(df)} rows")

def resolve_path(name):
    for ext in ['', '.jpg', '.jpeg', '.png']:
        p = os.path.join(ISIC_IMG_DIR, str(name) + ext)
        if os.path.exists(p):
            return p
    return None

def save_fig_both(fig, name, dpi=300):
    fig.savefig(f'{name}.png', dpi=dpi, bbox_inches='tight', facecolor='white')
    fig.savefig(f'{name}.pdf', bbox_inches='tight', facecolor='white')
    print(f"  Saved: {name}.png + {name}.pdf")


def show_concept_extremes(concept_name, n=6, target_filter=None, title_suffix=""):
    """Show the n highest and n lowest scoring real images for a concept,
    optionally restricted to one class (target_filter=1 melanoma, 0 benign)."""
    col = f'concept_{concept_name}'
    sub = df if target_filter is None else df[df['target'] == target_filter]

    top = sub.nlargest(n, col)
    bottom = sub.nsmallest(n, col)

    fig, axes = plt.subplots(2, n, figsize=(2.2*n, 5))
    fig.suptitle(f"{concept_name}{title_suffix} — real images, CBM-A predicted score",
                 fontweight='bold', fontsize=12)

    for row_idx, (subset, label) in enumerate([(top, 'HIGHEST'), (bottom, 'LOWEST')]):
        for i, (_, r) in enumerate(subset.iterrows()):
            ax = axes[row_idx, i]
            path = resolve_path(r['image_name'])
            if path:
                img = Image.open(path).convert('RGB')
                ax.imshow(img)
            else:
                ax.text(0.5, 0.5, 'not found', ha='center')
            diag = 'MEL' if r['target'] == 1 else 'BEN'
            ax.set_title(f"{r[col]:.3f} [{diag}]", fontsize=9)
            ax.set_xticks([]); ax.set_yticks([])
        axes[row_idx, 0].set_ylabel(label, fontsize=10, fontweight='bold')

    plt.tight_layout()
    save_fig_both(fig, f'nb2b_extremes_{concept_name}')
    plt.show()


# ============================================================
# THE KEY CHECK — blue_whitish_veil
# Top row: the highest-scoring images IN THE ENTIRE CORPUS (both classes,
# since within-melanoma top values are themselves near-floor — we need to
# see what CBM-A's absolute highest confidence looks like at all)
# ============================================================
print("\n" + "="*60)
print("CHECK 1: blue_whitish_veil — is the collapse visually real?")
print("="*60)
show_concept_extremes('blue_whitish_veil', n=6, target_filter=None,
                      title_suffix=" (top across ALL images, not just melanoma)")

# Also specifically among melanoma cases only — these are the highest BWV
# values melanoma images get, even though absolute scores are tiny
print("\nWithin melanoma cases only (absolute values will be small):")
show_concept_extremes('blue_whitish_veil', n=6, target_filter=1,
                      title_suffix=" (melanoma only)")


# ============================================================
# COMPARISON CHECK — dots_and_globules (the healthiest concept)
# for visual calibration: what does a WORKING concept look like?
# ============================================================
print("\n" + "="*60)
print("CHECK 2: dots_and_globules — comparison (this concept is healthy)")
print("="*60)
show_concept_extremes('dots_and_globules', n=6, target_filter=1,
                      title_suffix=" (melanoma only)")


# ============================================================
# COMPARISON CHECK — vascular_structures (bimodal / inflated vs Derm7pt)
# ============================================================
print("\n" + "="*60)
print("CHECK 3: vascular_structures — bimodal pattern, worth a look")
print("="*60)
show_concept_extremes('vascular_structures', n=6, target_filter=1,
                      title_suffix=" (melanoma only)")

print("\n" + "="*60)
print("Inspect the printed scores against the actual images:")
print("  - If BWV-'highest' images genuinely show no visible blue-white")
print("    haze, the collapse is a real absence in this corpus (proceed")
print("    with the documented limitation, no fix needed).")
print("  - If some clearly DO show visible blue-whitish veil but still")
print("    score near-zero, that's a real calibration failure worth a")
print("    closer look before committing to NB3.")
print("="*60)
```
